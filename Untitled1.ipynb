```python
import pandas as pd
import numpy as np

# Load and process data correctly
data = pd.read_csv('sample_data/mnist_train_small.csv', header=None)
data = data.to_numpy().T

# Row 0 contains labels, Rows 1 to 785 contain 784 pixels
data_Y = data[0, :]             # Shape: (20000,)
data_X = data[1:, :] / 255.0    # Shape: (784, 20000)

# Clean split
data_train_X = data_X[:, :10000]
data_train_Y = data_Y[:10000]
data_test_X = data_X[:, 10000:]
data_test_Y = data_Y[10000:]

def init_params():
  # Multiplied by 0.01 or 0.1 so initial values aren't too massive
  W1 = (np.random.rand(10, 784) - 0.5) * 0.1
  B1 = (np.random.rand(10, 1) - 0.5) * 0.1
  W2 = (np.random.rand(10, 10) - 0.5) * 0.1
  B2 = (np.random.rand(10, 1) - 0.5) * 0.1
  return W1, B1, W2, B2

def ReLU(M):
  return np.maximum(0, M)

def softmax(x):
  # Calculate vertically along columns (axis=0)
  e_x = np.exp(x - np.max(x, axis=0, keepdims=True))
  return e_x / np.sum(e_x, axis=0, keepdims=True)

def forward_prop(W1, B1, W2, B2, X):
  M1 = W1.dot(X) + B1
  A1 = ReLU(M1)
  M2 = W2.dot(A1) + B2
  A2 = softmax(M2)
  return A2, M1, A1

def one_hot(Y):
  num_classes = 10  # MNIST has exactly 10 classes (0-9)
  one_hot_Y = np.zeros((num_classes, Y.size), dtype=int)
  one_hot_Y[Y, np.arange(Y.size)] = 1
  return one_hot_Y

def back_prop(W1, W2, A2, A1, X, Y, M1):
  one_hot_Y = one_hot(Y)
  dM2 = A2 - one_hot_Y
  m = X.shape[1]

  dW2 = (1 / m) * np.dot(dM2, A1.T)  # Fixed: Use A1 instead of A2
  dB2 = (1 / m) * np.sum(dM2, axis=1, keepdims=True)

  g_prime_M1 = (M1 > 0).astype(float)
  dM1 = np.dot(W2.T, dM2) * g_prime_M1

  dW1 = (1 / m) * np.dot(dM1, X.T)
  dB1 = (1 / m) * np.sum(dM1, axis=1, keepdims=True)
  return dW2, dB2, dW1, dB1, one_hot_Y

def gradient_descent(W1, B1, W2, B2, X, Y, n, a):
  for i in range(n):
    A2, M1, A1 = forward_prop(W1, B1, W2, B2, X)
    dW2, dB2, dW1, dB1, one_hot_Y = back_prop(W1, W2, A2, A1, X, Y, M1)

    W2 -= a * dW2
    B2 -= a * dB2
    W1 -= a * dW1
    B1 -= a * dB1

    if i % 100 == 0:
      predictions = np.argmax(A2, axis=0)
      accuracy = np.mean(predictions == Y)
      print(f"Iteration {i}: Accuracy = {accuracy * 100:.2f}%")

  return W1, B1, W2, B2

# Train the network
W1, B1, W2, B2 = init_params()
W1, B1, W2, B2 = gradient_descent(W1, B1, W2, B2, data_train_X, data_train_Y, 2000, 0.1)
def check_test_accuracy(W1, B1, W2, B2, X_test, Y_test):
    # 1. Run a single forward pass with the test data
    A2, _, _ = forward_prop(W1, B1, W2, B2, X_test)

    # 2. Get predictions
    predictions = np.argmax(A2, axis=0)

    # 3. Calculate accuracy against the test labels
    test_acc = np.mean(predictions == Y_test)
    print(f"\nFinal Test Set Accuracy: {test_acc * 100:.2f}%")
    return predictions

# Run it using your trained parameters and test variables
test_predictions = check_test_accuracy(W1, B1, W2, B2, data_test_X, data_test_Y)
# Save your raw math matrices into a single file
np.savez('mnist_numpy_weights.npz', W1=W1, B1=B1, W2=W2, B2=B2)
print("💾 NumPy model weights successfully saved as 'mnist_numpy_weights.npz'!")
```

```python
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader

# Check for GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Training will run on: {device}")

# 1. Load data
data = pd.read_csv('sample_data/mnist_train_small.csv', header=None).to_numpy()
X_raw, Y_raw = data[:, 1:] / 255.0, data[:, 0]
X_imgs = X_raw.reshape(-1, 1, 28, 28)

X_train, Y_train = torch.FloatTensor(X_imgs[:10000]), torch.LongTensor(Y_raw[:10000])
X_test, Y_test = torch.FloatTensor(X_imgs[10000:]).to(device), torch.LongTensor(Y_raw[10000:]).to(device)

train_loader = DataLoader(TensorDataset(X_train, Y_train), batch_size=64, shuffle=True)

# 2. Build the CNN architecture
model = nn.Sequential(
    nn.Conv2d(in_channels=1, out_channels=16, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2, stride=2),
    nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1),
    nn.ReLU(),
    nn.MaxPool2d(kernel_size=2, stride=2),
    nn.Flatten(),
    nn.Linear(1568, 64),
    nn.ReLU(),
    nn.Linear(64, 10)
).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 3. Train for 25 epochs to maximize accuracy
epochs = 25
print("Starting training loop...")
for epoch in range(epochs):
    model.train()
    correct, total = 0, 0
    for batch_X, batch_Y in train_loader:
        batch_X, batch_Y = batch_X.to(device), batch_Y.to(device)
        outputs = model(batch_X)
        loss = criterion(outputs, batch_Y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        _, predicted = torch.max(outputs, 1)
        total += batch_Y.size(0)
        correct += (predicted == batch_Y).sum().item()

    if epoch % 5 == 0 or epoch == epochs - 1:
        print(f"Epoch {epoch:2d} | Train Accuracy: {(correct/total)*100:.2f}%")

# 4. Verify test split performance
model.eval()
with torch.no_grad():
    test_outputs = model(X_test)
    _, test_preds = torch.max(test_outputs, dim=1)
    test_accuracy = (test_preds == Y_test).float().mean() * 100
    print(f"\n🔥 Training Complete! Test Accuracy: {test_accuracy.item():.2f}%")

# 5. SAVE THE MODEL WEIGHTS TO DISK
torch.save(model.state_dict(), 'mnist_cnn.pt')
print("💾 Successfully saved model file as 'mnist_cnn.pt' in your sidebar explorer!")
```

```python
import pandas as pd
import numpy as np

# ==========================================
# 1. LOAD AND PROCESS DATA (18,000 IMAGES)
# ==========================================
print("Loading data...")
data = pd.read_csv('sample_data/mnist_train_small.csv', header=None)
data = data.to_numpy()

data_Y = data[:, 0]
data_X = data[:, 1:] / 255.0
data_X = data_X.T

# Split layout: 18,000 for training, 2,000 for validation testing
data_train_X = data_X[:, :18000].T.reshape(18000, 28, 28)
data_train_Y = data_Y[:18000]
data_test_X = data_X[:, 18000:].T.reshape(2000, 28, 28)
data_test_Y = data_Y[18000:]

def one_hot_encode(labels, num_classes=10):
    num_samples = labels.size
    one_hot = np.zeros((num_samples, num_classes))
    one_hot[np.arange(num_samples), labels.astype(int)] = 1
    return one_hot

data_train_Y_encoded = one_hot_encode(data_train_Y)
data_test_Y_encoded = one_hot_encode(data_test_Y)

data_train_X = np.expand_dims(data_train_X, axis=1) # Shape: (18000, 1, 28, 28)
data_test_X = np.expand_dims(data_test_X, axis=1)   # Shape: (2000, 1, 28, 28)

# ==========================================
# 2. INITIALIZATION & LEAKY ACTIVATIONS
# ==========================================
def init_filters(num_filters=16):
    input_channels = 1
    filter_size = 3
    d_in = input_channels * filter_size * filter_size
    std_dev = np.sqrt(2.0 / d_in)
    W1 = np.random.randn(num_filters, input_channels, filter_size, filter_size) * std_dev
    B1 = np.zeros((num_filters, 1))
    return W1, B1

def init_dense(input_nodes=2704, output_nodes=10):
    std_dev = np.sqrt(1.0 / input_nodes)
    W2 = np.random.randn(output_nodes, input_nodes) * std_dev
    B2 = np.zeros((output_nodes, 1))
    return W2, B2

W1, B1 = init_filters(16)
W2, B2 = init_dense(2704, 10)

# ADAM CONFIGURATION: Initialize first (m) and second (v) moment states to 0
m_W1, v_W1 = np.zeros_like(W1), np.zeros_like(W1)
m_B1, v_B1 = np.zeros_like(B1), np.zeros_like(B1)
m_W2, v_W2 = np.zeros_like(W2), np.zeros_like(W2)
m_B2, v_B2 = np.zeros_like(B2), np.zeros_like(B2)

t = 0

# Leaky ReLU fixes dead filter neurons permanently
def leaky_relu(Z, alpha=0.1):
    return np.where(Z > 0, Z, alpha * Z)

def leaky_relu_derivative(Z, alpha=0.1):
    return np.where(Z > 0, 1.0, alpha)

def softmax(Z):
    exp_Z = np.exp(Z - np.max(Z, axis=0, keepdims=True))
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)

def compute_loss(A2, Y_encoded):
    Y = Y_encoded.T
    A2_clipped = np.clip(A2, 1e-15, 1.0 - 1e-15)
    loss_per_sample = -np.sum(Y * np.log(A2_clipped), axis=0)
    return np.mean(loss_per_sample)

# ==========================================
# 3. VECTORIZED CONVOLUTION & POOL ENGINES
# ==========================================
def im2col_indices(x_shape, kh, kw, stride=1):
    N, C, H, W = x_shape
    out_h = (H - kh) // stride + 1
    out_w = (W - kw) // stride + 1
    i0 = np.repeat(np.arange(kh), kw)
    i0 = np.tile(i0, C)
    i1 = stride * np.repeat(np.arange(out_h), out_w)
    j0 = np.tile(np.arange(kw), kh * C)
    j1 = stride * np.tile(np.arange(out_w), out_h)
    i = i0.reshape(-1, 1) + i1.reshape(1, -1)
    j = j0.reshape(-1, 1) + j1.reshape(1, -1)
    k = np.repeat(np.arange(C), kh * kw).reshape(-1, 1)
    return k, i, j

def vectorized_convolution(X, W, b):
    n_filters, _, f_h, f_w = W.shape
    n_x, _, img_h, img_w = X.shape
    out_h, out_w = img_h - f_h + 1, img_w - f_w + 1
    k, i, j = im2col_indices(X.shape, f_h, f_w)
    X_cols = X[:, k, i, j].transpose(1, 2, 0).reshape(f_h * f_w, -1)
    Z = np.dot(W.reshape(n_filters, -1), X_cols) + b
    return Z.reshape(n_filters, out_h, out_w, n_x).transpose(3, 0, 1, 2)

def vectorized_max_pool(A, pool_size=2, stride=2):
    N, C, H, W = A.shape
    out_h, out_w = H // pool_size, W // pool_size
    A_reshaped = A.reshape(N, C, out_h, pool_size, out_w, pool_size)
    return A_reshaped.max(axis=(3, 5))

# ==========================================
# 4. BACKPROPAGATION & ADAM OPTIMIZER MATH
# ==========================================
def backward_dense(A2, Y_encoded, flattened_X, W2, l2_lambda=0.001):
    m = Y_encoded.shape[0]  # EXTRACTS SCALAR COUNT CORRECTLY
    Y = Y_encoded.T
    dZ2 = A2 - Y
    # Added L2 Regularization penalty to prevent dominating weights
    dW2 = (1 / m) * np.dot(dZ2, flattened_X.T) + (l2_lambda / m) * W2
    dB2 = (1 / m) * np.sum(dZ2, axis=1, keepdims=True)
    dFlattened = np.dot(W2.T, dZ2)
    return dZ2, dW2, dB2, dFlattened

def vectorized_backward_max_pool(dP1, A1, pool_size=2, stride=2):
    N, C, H, W = A1.shape
    out_h, out_w = H // pool_size, W // pool_size
    A1_reshaped = A1.reshape(N, C, out_h, pool_size, out_w, pool_size)
    out_max = A1_reshaped.max(axis=(3, 5), keepdims=True)
    mask = (A1_reshaped == out_max)
    dP1_reshaped = dP1.reshape(N, C, out_h, 1, out_w, 1)
    dA1_reshaped = dP1_reshaped * mask
    dA1 = dA1_reshaped.reshape(N, C, H, W)
    return dA1

def backward_convolution(X, dZ1, W1, l2_lambda=0.001):
    n_filters, d_filter, f_h, f_w = W1.shape
    m = X.shape[0]          # EXTRACTS SCALAR COUNT CORRECTLY
    dZ1_reshaped = dZ1.transpose(1, 2, 3, 0).reshape(n_filters, -1)
    k, i, j = im2col_indices(X.shape, f_h, f_w)
    X_cols = X[:, k, i, j].transpose(1, 2, 0).reshape(d_filter * f_h * f_w, -1)
    # Added L2 Regularization penalty to filters
    dW1_flat = np.dot(dZ1_reshaped, X_cols.T) / m + (l2_lambda / m) * W1.reshape(n_filters, -1)
    dW1 = dW1_flat.reshape(W1.shape)
    dB1 = np.sum(dZ1_reshaped, axis=1, keepdims=True) / m
    return dW1, dB1

def update_parameters_adam(W, B, dW, dB, m_W, v_W, m_B, v_B, t, lr=0.002, b1=0.9, b2=0.999, eps=1e-8):
    m_W = b1 * m_W + (1 - b1) * dW
    m_B = b1 * m_B + (1 - b1) * dB
    v_W = b2 * v_W + (1 - b2) * (dW ** 2)
    v_B = b2 * v_B + (1 - b2) * (dB ** 2)

    m_W_corrected = m_W / (1 - b1 ** t)
    m_B_corrected = m_B / (1 - b1 ** t)
    v_W_corrected = v_W / (1 - b2 ** t)
    v_B_corrected = v_B / (1 - b2 ** t)

    W -= lr * m_W_corrected / (np.sqrt(v_W_corrected) + eps)
    B -= lr * m_B_corrected / (np.sqrt(v_B_corrected) + eps)
    return W, B, m_W, v_W, m_B, v_B

# ==========================================
# 5. BALANCED TRAINING ENGINE
# ==========================================
epochs = 12
learning_rate = 0.002
decay_rate = 0.85
batch_size = 128
num_samples = 18000

print("Starting Balanced Leaky-ReLU Training Loop...")
print("-" * 60)

for epoch in range(epochs):
    permutation = np.random.permutation(num_samples)
    X_shuffled = data_train_X[permutation]
    Y_shuffled = data_train_Y_encoded[permutation]

    epoch_loss = 0
    epoch_acc = 0
    num_batches = num_samples // batch_size

    for b in range(num_batches):
        t += 1

        start_idx = b * batch_size
        end_idx = start_idx + batch_size

        X_batch = X_shuffled[start_idx:end_idx]
        Y_batch = Y_shuffled[start_idx:end_idx]

        # A. Forward Pass
        Z1 = vectorized_convolution(X_batch, W1, B1)
        A1 = leaky_relu(Z1)
        P1 = vectorized_max_pool(A1)
        flattened_X = P1.reshape(batch_size, -1).T

        Z2 = np.dot(W2, flattened_X) + B2
        A2 = softmax(Z2)

        # B. Metrics
        epoch_loss += compute_loss(A2, Y_batch)
        predictions = np.argmax(A2, axis=0)
        true_labels = np.argmax(Y_batch.T, axis=0)
        epoch_acc += np.mean(predictions == true_labels)

        # C. Backward Pass
        dZ2, dW2, dB2, dFlattened = backward_dense(A2, Y_batch, flattened_X, W2, l2_lambda=0.001)
        dP1 = dFlattened.T.reshape(batch_size, 16, 13, 13)
        dA1 = vectorized_backward_max_pool(dP1, A1)
        dZ1 = dA1 * leaky_relu_derivative(Z1)
        dW1, dB1 = backward_convolution(X_batch, dZ1, W1, l2_lambda=0.001)

        # D. Updates
        W1, B1, m_W1, v_W1, m_B1, v_B1 = update_parameters_adam(W1, B1, dW1, dB1, m_W1, v_W1, m_B1, v_B1, t, learning_rate)
        W2, B2, m_W2, v_W2, m_B2, v_B2 = update_parameters_adam(W2, B2, dW2, dB2, m_W2, v_W2, m_B2, v_B2, t, learning_rate)

    learning_rate = learning_rate * decay_rate
    avg_loss = epoch_loss / num_batches
    avg_acc = (epoch_acc / num_batches) * 100
    print(f"Epoch {epoch+1:02d}/{epochs} | Avg Loss: {avg_loss:.4f} | Avg Accuracy: {avg_acc:.2f}%")

print("-" * 60)
print("Training Completed Successfully!")
import pandas as pd
import numpy as np

# ==========================================
# 1. LOAD THE TEST DATA ONLY
# ==========================================
print("Loading MNIST data...")
data = pd.read_csv('sample_data/mnist_train_small.csv', header=None).to_numpy()

data_Y = data[:, 0]
data_X = data[:, 1:] / 255.0
data_X = data_X.T

# Use the exact same test slice your training script used (the final 2,000 images)
data_test_X = data_X[:, 18000:].T.reshape(2000, 28, 28)
data_test_Y = data_Y[18000:]

def one_hot_encode(labels, num_classes=10):
    num_samples = labels.size
    one_hot = np.zeros((num_samples, num_classes))
    one_hot[np.arange(num_samples), labels.astype(int)] = 1
    return one_hot

data_test_Y_encoded = one_hot_encode(data_test_Y)
data_test_X = np.expand_dims(data_test_X, axis=1)   # Shape: (2000, 1, 28, 28)

# ==========================================
# 2. LOAD YOUR TRAINED WEIGHTS
# ==========================================
print("Loading trained network weights...")
W1 = np.load('adam_cnn_W1.npy')
B1 = np.load('adam_cnn_B1.npy')
W2 = np.load('adam_cnn_W2.npy')
B2 = np.load('adam_cnn_B2.npy')

# ==========================================
# 3. MATH PASS FUNCTIONS
# ==========================================
def leaky_relu(Z, alpha=0.1):
    return np.where(Z > 0, Z, alpha * Z)

def softmax(Z):
    exp_Z = np.exp(Z - np.max(Z, axis=0, keepdims=True))
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)

def im2col_indices(x_shape, kh, kw, stride=1):
    N, C, H, W = x_shape
    out_h = (H - kh) // stride + 1
    out_w = (W - kw) // stride + 1
    i0 = np.repeat(np.arange(kh), kw)
    i0 = np.tile(i0, C)
    i1 = stride * np.repeat(np.arange(out_h), out_w)
    j0 = np.tile(np.arange(kw), kh * C)
    j1 = stride * np.tile(np.arange(out_w), out_h)
    i = i0.reshape(-1, 1) + i1.reshape(1, -1)
    j = j0.reshape(-1, 1) + j1.reshape(1, -1)
    k = np.repeat(np.arange(C), kh * kw).reshape(-1, 1)
    return k, i, j

def vectorized_convolution(X, W, b):
    n_filters, _, f_h, f_w = W.shape
    n_x, _, img_h, img_w = X.shape
    out_h, out_w = img_h - f_h + 1, img_w - f_w + 1
    k, i, j = im2col_indices(X.shape, f_h, f_w)
    X_cols = X[:, k, i, j].transpose(1, 2, 0).reshape(f_h * f_w, -1)
    Z = np.dot(W.reshape(n_filters, -1), X_cols) + b
    return Z.reshape(n_filters, out_h, out_w, n_x).transpose(3, 0, 1, 2)

def vectorized_max_pool(A, pool_size=2, stride=2):
    N, C, H, W = A.shape
    out_h, out_w = H // pool_size, W // pool_size
    A_reshaped = A.reshape(N, C, out_h, pool_size, out_w, pool_size)
    return A_reshaped.max(axis=(3, 5))

# ==========================================
# 4. RUN INFERENCE PASS ON TEST DATA
# ==========================================
print("Computing predictions on unseen data...")
Z1_test = vectorized_convolution(data_test_X, W1, B1)
A1_test = leaky_relu(Z1_test) # Match your Leaky-ReLU training upgrade
P1_test = vectorized_max_pool(A1_test)
flattened_X_test = P1_test.reshape(2000, -1).T

Z2_test = np.dot(W2, flattened_X_test) + B2
A2_test = softmax(Z2_test)

# Calculate testing accuracy score
test_predictions = np.argmax(A2_test, axis=0)
test_true_labels = np.argmax(data_test_Y_encoded.T, axis=0)
test_accuracy = np.mean(test_predictions == test_true_labels) * 100

print("-" * 50)
print(f"✅ UNSEEN TEST DATASET ACCURACY: {test_accuracy:.2f}%")
print("-" * 50)

# Save final stable arrays
np.save('adam_cnn_W1.npy', W1)
np.save('adam_cnn_B1.npy', B1)
np.save('adam_cnn_W2.npy', W2)
np.save('adam_cnn_B2.npy', B2)
print("Stable parameters saved successfully!")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import io

# ==========================================
# 1. HANDLE IMAGE UPLOAD (Jupyter/Colab vs Local)
# ==========================================
try:
    from google.colab import files
    print("Please select your handwritten image file:")
    uploaded_files = files.upload()
    file_name = list(uploaded_files.keys())[0]
    img_data = io.BytesIO(uploaded_files[file_name])
except ImportError:
    # If running locally on your computer, modify this path string to your image
    file_name = "my_handwritten_digit.png"
    img_data = file_name

# ==========================================
# 2. LOAD PRE-TRAINED PARAMETERS INSTANTLY
# ==========================================
print("\nLoading pre-trained 97.30% Leaky-ReLU CNN parameters...")
try:
    W1 = np.load('adam_cnn_W1.npy')
    B1 = np.load('adam_cnn_B1.npy')
    W2 = np.load('adam_cnn_W2.npy')
    B2 = np.load('adam_cnn_B2.npy')
except FileNotFoundError:
    print("❌ Error: Weight files not found. Ensure you ran the new training script first!")
    raise

# ==========================================
# 3. MATH ENGINE OPERATIONS (WITH LEAKY RELU)
# ==========================================
# MATCHES TRAINING MATH PERMANENTLY
def leaky_relu(Z, alpha=0.1):
    return np.where(Z > 0, Z, alpha * Z)

def softmax(Z):
    exp_Z = np.exp(Z - np.max(Z, axis=0, keepdims=True))
    return exp_Z / np.sum(exp_Z, axis=0, keepdims=True)

def im2col_indices(x_shape, kh, kw, stride=1):
    N, C, H, W = x_shape
    out_h = (H - kh) // stride + 1
    out_w = (W - kw) // stride + 1
    i0 = np.repeat(np.arange(kh), kw)
    i0 = np.tile(i0, C)
    i1 = stride * np.repeat(np.arange(out_h), out_w)
    j0 = np.tile(np.arange(kw), kh * C)
    j1 = stride * np.tile(np.arange(out_w), out_h)
    i = i0.reshape(-1, 1) + i1.reshape(1, -1)
    j = j0.reshape(-1, 1) + j1.reshape(1, -1)
    k = np.repeat(np.arange(C), kh * kw).reshape(-1, 1)
    return k, i, j

def vectorized_convolution(X, W, b):
    n_filters, _, f_h, f_w = W.shape
    n_x, _, img_h, img_w = X.shape
    out_h, out_w = img_h - f_h + 1, img_w - f_w + 1
    k, i, j = im2col_indices(X.shape, f_h, f_w)
    X_cols = X[:, k, i, j].transpose(1, 2, 0).reshape(f_h * f_w, -1)
    Z = np.dot(W.reshape(n_filters, -1), X_cols) + b
    return Z.reshape(n_filters, out_h, out_w, n_x).transpose(3, 0, 1, 2)

def vectorized_max_pool(A, pool_size=2, stride=2):
    N, C, H, W = A.shape
    out_h, out_w = H // pool_size, W // pool_size
    A_reshaped = A.reshape(N, C, out_h, pool_size, out_w, pool_size)
    return A_reshaped.max(axis=(3, 5))

def deskew_single_image(img_2d):
    """Straightens out skewed shapes using pixel covariance metrics."""
    H, W = img_2d.shape
    y, x = np.indices((H, W))

    total_mass = np.sum(img_2d)
    if total_mass == 0:
        return img_2d

    cx = np.sum(x * img_2d) / total_mass
    cy = np.sum(y * img_2d) / total_mass

    x_shifted = x - cx
    y_shifted = y - cy

    mu_11 = np.sum(x_shifted * y_shifted * img_2d) / total_mass
    mu_02 = np.sum(y_shifted * y_shifted * img_2d) / total_mass

    if mu_02 < 1e-6:
        return img_2d

    skew = mu_11 / mu_02

    img_deskewed = np.zeros_like(img_2d)
    for i in range(H):
        for j in range(W):
            new_j = int(np.round(j - skew * (i - cy)))
            if 0 <= new_j < W:
                img_deskewed[i, new_j] = img_2d[i, j]

    return img_deskewed

# ==========================================================
# 4. FIXED IMAGE PREPROCESSING (MNIST CENTERING & DESKEW)
# ==========================================================
print(f"Processing image with MNIST-grade alignment: {file_name}...")

# 1. Open image and convert to grayscale
raw_img = Image.open(img_data).convert('L')

# 2. Convert to numpy array and scale pixels
img_np = np.array(raw_img, dtype=np.float32) / 255.0

# 3. Correct dark brush strokes on light background
if np.mean(img_np) > 0.5:
    img_np = 1.0 - img_np

# 4. Crop the canvas tightly to ONLY the handwritten strokes
# This completely eliminates canvas aspect-ratio distortions!
rows = np.any(img_np > 0.1, axis=1)
cols = np.any(img_np > 0.1, axis=0)
if not np.any(rows) or not np.any(cols):
    # Fallback for empty image
    custom_X_final = np.zeros((28, 28))
else:
    ymin, ymax = np.where(rows)[0][0], np.where(rows)[0][-1]
    xmin, xmax = np.where(cols)[0][0], np.where(cols)[0][-1]
    cropped_digit = img_np[ymin:ymax+1, xmin:xmax+1]

    # 5. Place the cropped digit into a perfect 20x20 box (MNIST Standard)
    # This preserves the original aspect ratio perfectly
    digit_pil = Image.fromarray((cropped_digit * 255).astype(np.uint8))
    h_c, w_c = cropped_digit.shape
    if w_c > h_c:
        new_w = 20
        new_h = int(20 * (h_c / w_c))
    else:
        new_h = 20
        new_w = int(20 * (w_c / h_c))
    # Ensure dimensions are at least 1
    new_w, new_h = max(1, new_w), max(1, new_h)
    digit_resized = digit_pil.resize((new_w, new_h), Image.Resampling.BILINEAR)

    # 6. Center that 20x20 digit inside a clean 28x28 background matrix
    # This completely eliminates translation shift errors!
    custom_X_final = np.zeros((28, 28))
    start_x = (28 - new_w) // 2
    start_y = (28 - new_h) // 2
    custom_X_final[start_y:start_y+new_h, start_x:start_x+new_w] = np.array(digit_resized, dtype=np.float32) / 255.0

# 7. Apply the deskewing math onto the perfectly centered frame
custom_X_deskewed = deskew_single_image(custom_X_final)

# Format to standard 4D tensor matrix layout (1, 1, 28, 28)
custom_X = np.expand_dims(custom_X_deskewed, axis=(0, 1))


# ==========================================
# 5. FORWARD INFERENCE PIPELINE
# ==========================================
Z1 = vectorized_convolution(custom_X, W1, B1)
A1 = leaky_relu(Z1)  # FIXED: Now matches Leaky ReLU training math completely!
P1 = vectorized_max_pool(A1)
flattened_X = P1.reshape(1, -1).T

Z2 = np.dot(W2, flattened_X) + B2
A2 = softmax(Z2)

predicted_digit = np.argmax(A2, axis=0)[0]
confidence = A2[predicted_digit, 0] * 100

print("-" * 50)
print(f"🎯 NETWORK PREDICTION RESULT: {predicted_digit}")
print(f"📈 CONFIDENCE RATING ESTIMATE: {confidence:.2f}%")
print("-" * 50)


# Figure 2: The 16 Activated Filter Feature Maps Grid Pass
feature_maps = A1[0]  # Shape becomes (16, 26, 26)
fig, axes = plt.subplots(4, 4, figsize=(8, 8))
fig.suptitle(f"Layer 1 Leaky-ReLU Feature Maps (Predicted as Digit: {predicted_digit})", fontsize=14, y=0.98)

for f in range(16):
    ax = axes[f // 4, f % 4]
    ax.imshow(feature_maps[f], cmap='gray')
    ax.axis('off')
    ax.set_title(f"Filter {f+1}", fontsize=10)

plt.tight_layout()
plt.show()
```
